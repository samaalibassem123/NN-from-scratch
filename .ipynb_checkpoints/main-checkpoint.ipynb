```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class NeuralNetwork:
    def __init__(self, X ,Y, inputs, outputs,hlayer_d):
        self.X , self.Y= X, Y
        self.w1 = np.random.randn(inputs,hlayer_d)
        self.w2 = np.random.randn(hlayer_d, outputs)
        self.b1 = np.zeros((1,hlayer_d))
        self.b2 = np.zeros((1,outputs))

    def forward(self, x):
        self.z1 = np.dot(x, self.w1) + self.b1
        self.a1 = self.Relu(self.z1)
        self.z2 = np.dot(self.a1, self.w2) + self.b2
        self.a2 = self.softmax(self.z2)
   

   
    
    def backward(self, alpha):
        #calculate the gradients 
        m = self.X.shape[1]
        one_hot_Y = self.one_hot(self.Y)
        
        dZ2 = self.a2 - one_hot_Y

        dW2 = (self.a1.T @ dZ2) / m
        db2 = np.sum(dZ2, axis=0, keepdims=True) / m

        # Hidden layer gradients
        dZ1 = (dZ2 @ self.w2.T) * self.Relu(self.z1, deriv=True)

        dW1 = (self.X.T @ dZ1) / m
        db1 = np.sum(dZ1, axis=0, keepdims=True) 

        # update paramters
        
        self.w1 = self.w1 - alpha * dW1
        self.b1 = self.b1 - alpha * db1    
        self.w2 = self.w2 - alpha * dW2  
        self.b2 = self.b2 - alpha * db2  


    
    def train(self, epochs,alpha=0.1):
        losses = []
        for i in range(epochs):
            self.forward(self.X)
            print(self.X)
            print("============ WEIGHTS ===========")
            print(self.w1)
            print(self.b1)
            print("========= Hidden LAYER =======")
            print(self.a1)
            print("============ WEIGHTS ===========")
            print(self.w2)
            print(self.b2)
            print("========= OUTPUT LAYER =======")
            print(self.a2)

            
            pred = self.get_predictions(self.a2)
            acc = self.get_accuracy(pred)
            
            print("Final predictions:", pred)
            print("loss:", self.loss())
            losses.append(self.loss())
            # Plot the loss curve
            plt.figure(figsize=(8, 5))
            plt.plot(range(1, len(losses) + 1), losses, label="Training Loss")
        
            plt.xlabel("Epoch")
            plt.ylabel("Cross-Entropy Loss")
            plt.title("Training Loss over Epochs")
            plt.grid(True)
            plt.legend()
            plt.tight_layout()
            plt.show()
            print(f"Final accuracy: {acc:.4%}")
            
            self.backward(alpha)

    def fit(self, value):
        self.forward(value)
        pred = self.get_predictions(self.a2)
        print("Final predictions:", pred)



     def Relu(self, A, deriv=False):
        if deriv:
            return (A > 0).astype(float)
    
        return np.maximum(0, A)
    

    def softmax(self, Z):
        Z = Z - np.max(Z, axis=1, keepdims=True)
        exp_Z = np.exp(Z)
        return exp_Z / np.sum(exp_Z, axis=1, keepdims=True)


    def one_hot(self,Y):
        one_hot_Y = np.zeros((Y.size, Y.max() + 1))
        one_hot_Y[np.arange(Y.size), Y] = 1
        one_hot_Y = one_hot_Y
        return one_hot_Y

        def loss(self):
        m = self.Y.size
        one_hot_Y = self.one_hot(self.Y)
    
        probabilities = np.clip(self.a2, 1e-12, 1.0)
    
        return -np.sum(one_hot_Y * np.log(probabilities)) / m
        
    def get_predictions(self, a2):
        return np.argmax(a2, axis=1)
    
    def get_accuracy(self, predictions):
        print("acc", predictions, self.Y)
        return round(np.sum(predictions == self.Y) / self.Y.size, 4)

        
        
    
        
```

# Train the NN

```python
X = np.array([
    [1.0, 2.0, 3.0],
    [2.0, 1.0, 4.0],
    [3.0, 2.0, 1.0],
    [4.0, 3.0, 2.0]
])

Y = np.array([1, 0, 1, 1])

nn = NeuralNetwork(
    X=X,
    Y=Y,
    inputs=3,
    outputs=2,
    hlayer_d=5
)

nn.train(epochs=100, alpha=0.01)
```
